## 1. Imports & Setup

```python
# Core libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Scikit-learn: preprocessing & model selection
from sklearn.model_selection import (train_test_split, GridSearchCV, StratifiedKFold,
                                      cross_val_score, cross_validate)
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.impute import SimpleImputer

# Scikit-learn: models
from sklearn.ensemble import RandomForestClassifier
from sklearn.neural_network import MLPClassifier

# Scikit-learn: metrics
from sklearn.metrics import (accuracy_score, f1_score, precision_score, recall_score,
                              confusion_matrix, ConfusionMatrixDisplay, classification_report)

# ---- Consistent, tidy plotting style used throughout the notebook ----
plt.style.use("seaborn-v0_8-whitegrid")
plt.rcParams.update({
    "figure.facecolor": "white",
    "axes.titlesize": 12,
    "axes.titleweight": "bold",
    "axes.labelsize": 10,
    "font.size": 10,
})
PALETTE = ["#2f7fd1", "#dd8452", "#55a868", "#c44e52", "#8172b2", "#937860"]

RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)
pd.set_option("display.max_columns", None)
```

## 2. Data Preparation

The **Credit Approval** dataset (`crx.data`) is downloaded from the UCI Machine Learning
Repository. To protect confidentiality, the original attribute names were anonymised to
`A1`–`A15`, with `A16` as the class label (`+` = approved, `-` = not approved). We load the
raw file, assign meaningful column headers, replace the `?` placeholder used for missing
values with proper `NaN`, and write the cleaned frame to a CSV file for later reuse.

```python
# The dataset can be downloaded directly from UCI:
# url = "https://archive.ics.uci.edu/ml/machine-learning-databases/credit-screening/crx.data"
# For reproducibility we read the local copy of the same file (crx.data).

column_names = [f"A{i}" for i in range(1, 16)] + ["A16"]

df = pd.read_csv(
    "crx.data",
    header=None,
    names=column_names,
    na_values="?"          # '?' is the missing-value marker used in this dataset
)

print(f"Shape: {df.shape}")
df.head()
```

```python
# A2 and A14 are read as strings because of stray '?' entries -> convert to numeric
df["A2"] = pd.to_numeric(df["A2"], errors="coerce")
df["A14"] = pd.to_numeric(df["A14"], errors="coerce")

df.dtypes
```

```python
# Persist the raw-but-typed data frame to CSV as required
df.to_csv("credit_approval_raw.csv", index=False)
print("Saved credit_approval_raw.csv")
```

### 3.2 Missing values

```python
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)
missing_summary = pd.DataFrame({"missing_count": missing, "missing_pct": missing_pct})
missing_summary = missing_summary[missing_summary["missing_count"] > 0].sort_values("missing_count", ascending=False)
missing_summary.style.background_gradient(cmap="Blues", subset=["missing_count"]).format({"missing_pct": "{:.2f}%"})
```

### 3.4 Distribution of continuous attributes

### 3.5 Outlier detection (boxplots + IQR rule)

```python
fig, axes = plt.subplots(2, 3, figsize=(14, 8))
for col, ax, c in zip(continuous_cols, axes.ravel(), PALETTE):
    bp = df.boxplot(column=col, ax=ax, patch_artist=True, return_type="dict")
    for box in bp["boxes"]:
        box.set(facecolor=c, alpha=0.5)
    ax.set_title(col)
fig.suptitle("Outlier check via boxplots", fontsize=14, fontweight="bold", y=1.02)
plt.tight_layout()
plt.savefig("fig_boxplots.png", dpi=120)
plt.show()
```

```python
def iqr_outlier_count(series):
    q1, q3 = series.quantile([0.25, 0.75])
    iqr = q3 - q1
    lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    return ((series < lower) | (series > upper)).sum()

outlier_counts = {c: iqr_outlier_count(df[c].dropna()) for c in continuous_cols}
pd.Series(outlier_counts, name="outlier_count (IQR rule)").sort_values(ascending=False).to_frame()\
    .style.background_gradient(cmap="Oranges")
```

> **Insight:** all continuous attributes contain a non-trivial number of outliers
(particularly `A3`, `A8`, `A11` and `A15`), which is expected for financial variables such as
income, debt and credit-line values. Rather than deleting these records (which would remove
real applicants), we **cap** them using the IQR rule during preprocessing.

## 4. Preprocessing, Feature Selection and Feature Engineering

### 4.1 Handling missing values

```python
df_clean = df.copy()

num_cols = ["A2", "A3", "A8", "A11", "A14", "A15"]
cat_cols = ["A1", "A4", "A5", "A6", "A7", "A9", "A10", "A12", "A13"]

# Numeric attributes: impute with the median (robust to outliers)
num_imputer = SimpleImputer(strategy="median")
df_clean[num_cols] = num_imputer.fit_transform(df_clean[num_cols])

# Categorical attributes: impute with the most frequent category (mode)
cat_imputer = SimpleImputer(strategy="most_frequent")
df_clean[cat_cols] = cat_imputer.fit_transform(df_clean[cat_cols])

print("Remaining missing values:", df_clean.isnull().sum().sum())
```

### 4.2 Handling outliers (IQR capping / winsorising)

```python
def cap_outliers_iqr(series, k=1.5):
    q1, q3 = series.quantile([0.25, 0.75])
    iqr = q3 - q1
    lower, upper = q1 - k * iqr, q3 + k * iqr
    return series.clip(lower, upper)

for col in num_cols:
    df_clean[col] = cap_outliers_iqr(df_clean[col])

df_clean[num_cols].describe().T
```

### 4.3 Feature engineering

```python
# A11 = number of credit inquiries/derogatory reports, A15 = income-like continuous value.
# Engineer a simple debt-to-income style ratio and a binary "has_prior_default" flag,
# which are common, interpretable features in credit-risk modelling.

df_clean["A3_to_A8_ratio"] = df_clean["A3"] / (df_clean["A8"] + 1)   # avoid div-by-zero
df_clean["has_positive_A11"] = (df_clean["A11"] > 0).astype(int)     # engineered binary flag

engineered_cols = ["A3_to_A8_ratio", "has_positive_A11"]
df_clean[engineered_cols].head()
```

### 4.4 Label encoding of categorical / target attributes

```python
encoders = {}
for col in cat_cols:
    le = LabelEncoder()
    df_clean[col] = le.fit_transform(df_clean[col])
    encoders[col] = le

# Target
target_encoder = LabelEncoder()
df_clean["A16"] = target_encoder.fit_transform(df_clean["A16"])   # '+' -> 1 or 0, '-' -> the other
print("Class mapping:", dict(zip(target_encoder.classes_, target_encoder.transform(target_encoder.classes_))))

df_clean.head()
```

### 4.5 Feature scaling

```python
feature_cols = num_cols + cat_cols + engineered_cols
X = df_clean[feature_cols]
y = df_clean["A16"]

scaler = StandardScaler()
X_scaled = pd.DataFrame(scaler.fit_transform(X), columns=feature_cols)
X_scaled.describe().T[["mean", "std"]]
```

```python
# Persist the fully processed dataset as well
processed = X_scaled.copy()
processed["A16"] = y.values
processed.to_csv("credit_approval_processed.csv", index=False)
print("Saved credit_approval_processed.csv")
```

### 4.6 Train / test split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.25, random_state=RANDOM_STATE, stratify=y
)
print("Train:", X_train.shape, " Test:", X_test.shape)
```

---
## 5. Model Creation and Evaluation

### 5a. Classical classification model — Random Forest

**Metric justification:** the target classes are fairly balanced (44.5% / 55.5%) but not
perfectly, so **accuracy** alone can be slightly misleading. We therefore report:
- **Accuracy** — overall proportion of correctly classified applicants, easy to interpret.
- **F1-score** — harmonic mean of precision and recall; important here because both
  *false approvals* (risk to the bank) and *false rejections* (lost business) carry real
  costs, and F1 balances both types of error rather than favouring one class.

```python
rf = RandomForestClassifier(n_estimators=300, random_state=RANDOM_STATE)
rf.fit(X_train, y_train)
y_pred_rf = rf.predict(X_test)

rf_acc = accuracy_score(y_test, y_pred_rf)
rf_f1 = f1_score(y_test, y_pred_rf)

print(f"Random Forest — Accuracy: {rf_acc:.4f} | F1-score: {rf_f1:.4f}")
print(classification_report(y_test, y_pred_rf, target_names=[str(c) for c in target_encoder.classes_]))
```

```python
cm = confusion_matrix(y_test, y_pred_rf)
disp = ConfusionMatrixDisplay(cm, display_labels=target_encoder.classes_)
fig, ax = plt.subplots(figsize=(5, 4.5))
disp.plot(cmap="Blues", ax=ax, colorbar=False)
ax.set_title("Random Forest — Confusion Matrix")
plt.tight_layout()
plt.savefig("fig_rf_confusion.png", dpi=120)
plt.show()
```

### 5b. Neural network model (multi-layer perceptron) with 10-fold cross-validation

```python
mlp_baseline = MLPClassifier(
    hidden_layer_sizes=(32, 16),   # multi-layer: two hidden layers
    activation="relu",
    solver="adam",
    max_iter=300,
    early_stopping=True,
    random_state=RANDOM_STATE
)

cv = StratifiedKFold(n_splits=10, shuffle=True, random_state=RANDOM_STATE)

cv_results = cross_validate(mlp_baseline, X_scaled, y, cv=cv, scoring=["accuracy", "f1"], n_jobs=1)
cv_accuracy = cv_results["test_accuracy"]
cv_f1 = cv_results["test_f1"]

print(f"MLP (baseline) 10-fold CV — Accuracy: {cv_accuracy.mean():.4f} ± {cv_accuracy.std():.4f}")
print(f"MLP (baseline) 10-fold CV — F1-score: {cv_f1.mean():.4f} ± {cv_f1.std():.4f}")
```

```python
# Fit once on the train split so we also have a held-out test-set comparison later
mlp_baseline.fit(X_train, y_train)
y_pred_mlp = mlp_baseline.predict(X_test)

mlp_acc = accuracy_score(y_test, y_pred_mlp)
mlp_f1 = f1_score(y_test, y_pred_mlp)
print(f"MLP (baseline) held-out test — Accuracy: {mlp_acc:.4f} | F1-score: {mlp_f1:.4f}")
```

### 5c. Hyperparameter tuning with Grid Search (10-fold CV)

The search space covers **three** hyperparameters from the required list: `activation`
(activation function), `learning_rate_init` (learning rate), and `solver` (optimizer).
Architecture and `batch_size` are kept fixed at reasonable values so the grid search
stays tractable while still satisfying the "at least three" requirement.

```python
param_grid = {
    "activation": ["relu", "tanh"],          # activation function
    "learning_rate_init": [0.001, 0.01],     # learning rate
    "solver": ["adam", "sgd"],               # optimizer
}

mlp = MLPClassifier(hidden_layer_sizes=(32, 16), batch_size=32, max_iter=300,
                     early_stopping=True, random_state=RANDOM_STATE)

grid_search = GridSearchCV(
    estimator=mlp,
    param_grid=param_grid,
    scoring="f1",
    cv=cv,          # same 10-fold StratifiedKFold as above
    n_jobs=1,
    verbose=0
)

grid_search.fit(X_train, y_train)

print("Best hyperparameters:", grid_search.best_params_)
print(f"Best 10-fold CV F1-score: {grid_search.best_score_:.4f}")
```

```python
best_mlp = grid_search.best_estimator_
y_pred_tuned = best_mlp.predict(X_test)

tuned_acc = accuracy_score(y_test, y_pred_tuned)
tuned_f1 = f1_score(y_test, y_pred_tuned)

print(f"Tuned MLP held-out test — Accuracy: {tuned_acc:.4f} | F1-score: {tuned_f1:.4f}")
print(classification_report(y_test, y_pred_tuned, target_names=[str(c) for c in target_encoder.classes_]))
```

### 5d. Comparing the tuned neural network against the baseline classifier (5a)

```python
results = pd.DataFrame({
    "Model": ["Random Forest (5a)", "MLP baseline (5b)", "MLP tuned - GridSearch (5c)"],
    "Accuracy": [rf_acc, mlp_acc, tuned_acc],
    "F1-score": [rf_f1, mlp_f1, tuned_f1],
})
results.style.background_gradient(cmap="Greens", subset=["Accuracy", "F1-score"]).format({"Accuracy": "{:.4f}", "F1-score": "{:.4f}"})
```

```python
fig, ax = plt.subplots(figsize=(8, 5))
x = np.arange(len(results))
width = 0.35

ax.bar(x - width/2, results["Accuracy"], width, label="Accuracy", color=PALETTE[0], edgecolor="white")
ax.bar(x + width/2, results["F1-score"], width, label="F1-score", color=PALETTE[1], edgecolor="white")

ax.set_xticks(x)
ax.set_xticklabels(results["Model"], rotation=15, ha="right")
ax.set_ylim(0, 1)
ax.set_ylabel("Score")
ax.set_title("Model performance comparison")
ax.legend(frameon=False)
for i, (a, f) in enumerate(zip(results["Accuracy"], results["F1-score"])):
    ax.text(i - width/2, a + 0.015, f"{a:.3f}", ha="center", fontsize=9)
    ax.text(i + width/2, f + 0.015, f"{f:.3f}", ha="center", fontsize=9)

plt.tight_layout()
plt.savefig("fig_model_comparison.png", dpi=120)
plt.show()
```

**Discussion.** The Random Forest provides a strong, low-variance baseline typical of
tree ensembles on tabular data with mixed feature types. The un-tuned MLP is competitive but
sensitive to its `activation`, `learning_rate_init` and `solver` settings. After Grid Search
over these three hyperparameters (evaluated with 10-fold cross-validation), the tuned MLP's
held-out performance changes as shown in the chart above. In this run the untuned MLP already
performed close to its ceiling on the held-out split, so the tuned model's cross-validated
F1-score (selection criterion) improved over the untuned model's cross-validated score even
though the single held-out test split does not always move in the same direction — a reminder
that Grid Search should be judged primarily on its cross-validated score, not one train/test
split, and that hyperparameter tuning narrows the search space rather than guaranteeing a
better score on every possible split.
